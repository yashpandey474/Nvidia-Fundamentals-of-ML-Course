```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
y_true = np.array([0, 0, 1, 1, 0, 1, 0, 1]) # expected
y_pred = np.array([0, 1, 1, 1, 0, 0, 0, 1]) # predicted
```

```python
accuracy_score(y_true, y_pred), precision_score(y_true, y_pred), recall_score(y_true, y_pred), f1_score(y_true, y_pred)
```

```python
confusion_matrix(y_true, y_pred)
```

```python
# Regression metrics
y_true_regression = np.array([3.0, -0.5, 2.0, 7.0])
y_pred_regression = np.array([2.5, 0.0, 2.0, 8.0])
```

```python
mean_squared_error(y_true_regression, y_pred_regression), mean_absolute_error(y_true_regression, y_pred_regression), r2_score(y_true_regression, y_pred_regression)
```

```python
rmse = np.sqrt(mean_squared_error(y_true_regression, y_pred_regression))
```
